```python
tasks = []
```

```python
def show_menu():
  print('\n-- TO-DO APP MENU')
  print('1. Add Task')
  print('2. View Tasks')
  print('3. Mark Task as completed')
  print('4. Delete Task')
  print('5. Exit')
```

```python
def add_task():
  task_name = input("Enter task name: ")
  task = {
      "task": task_name,
      "completed": False
  }
  tasks.append(task)
  print("Task added successfully!")
```

```python
def view_tasks():
  if len(tasks) == 0:
    print("No tasks available.")
    return
  print("\nYour Tasks")
  for index, task in enumerate(tasks,start=1):
    status = "Done" if tasks["completed"] else "pending"
    print(index,".", tasks["task"], "-", status)
```

```python
def complete_task():
  view_tasks()
  if len(tasks) == 0:
    return
  try:
    task_no = int(input("Enter task number to mark as completed"))
    tasks[task_no -1]["completed"] = True
    print("Task marked as completed")
  except:
    print("Invalid Input")
```

```python
def delete_task():
  view_tasks()
  if len(tasks) ==0:
    return

  try:
    task_no = int(input("Enter task number to delete: "))
    tasks.pop(task_no -1)
  except:
    print("Invalid Input")
```

```python
def view_tasks():
  if len(tasks) == 0:
    print("No tasks available.")
    return
  print("\nYour Tasks")
  for index, task_item in enumerate(tasks,start=1):
    status = "Done" if task_item["completed"] else "pending"
    print(index,".", task_item["task"], "-", status)

while True:
  show_menu()
  choice = input('Enter your choice (1-5): ')

  if choice == "1":
    add_task()
  elif choice == "2":
    view_tasks()
  elif choice == "3":
    complete_task()
  elif choice == "4":
    delete_task()
  elif choice == "5":
    break
```
